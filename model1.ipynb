```python
import numpy as np
from sklearn.metrics import accuracy_score, precision_score, recall_score

# Suppose you already have:
# y_true = list of ground-truth labels (0 = different person, 1 = same person)
# y_scores = similarity scores from your face recognition model

def evaluate_thresholds(y_true, y_scores):
    thresholds = np.arange(0.5, 0.9, 0.02)  # sweep from 0.50 to 0.88
    print("Threshold | Accuracy | Precision | Recall")
    print("------------------------------------------")
    for t in thresholds:
        y_pred = (y_scores >= t).astype(int)
        acc = accuracy_score(y_true, y_pred)
        prec = precision_score(y_true, y_pred)
        rec = recall_score(y_true, y_pred)
        print(f"{t:.2f}      | {acc:.3f}   | {prec:.3f}    | {rec:.3f}")

# Example usage:
# evaluate_thresholds(y_true, y_scores)
```

```python
import cv2
import numpy as np

def get_embedding(path):
    img = cv2.imread(path)

    faces = app.get(img)

    if not faces:
        return None

    face = max(
        faces,
        key=lambda f: (f.bbox[2]-f.bbox[0]) *
                      (f.bbox[3]-f.bbox[1])
    )

    emb = face.embedding
    return emb / np.linalg.norm(emb)
```
